```python
import pandas as pd
import numpy as np
import xgboost as xgb
import matplotlib.pyplot as plt
from tqdm import tqdm
import re
from sklearn.metrics import mean_absolute_error
import warnings

warnings.filterwarnings('ignore')


# 1. --- CONFIGURATION ---
FUNDAMENTAL_PATH = r"SP500_Fundamental_History_Raw.csv"
PIT_PRICE_PATH = r"SP500_FINAL_BACKTEST_READY.csv"
# 2. --- DATA LOADING & CLEANING ---
print("Loading Data...")
df_funda = pd.read_csv(FUNDAMENTAL_PATH)
df_funda['Ticker'] = df_funda['Ticker'].apply(clean_ticker)
df_funda = df_funda.drop_duplicates(subset=['Ticker', 'Year'], keep='last')

df_prices = pd.read_csv(PIT_PRICE_PATH, index_col=0)
df_prices.index = df_prices.index.astype(int)
print("Fundamentals (Fiscal Year)")
print(df_funda[["Year", "Ticker", "Revenue", "Gross Profit"]].head(3))

print("Prices (1st of April)")
print(df_prices[["AAPL", "MSFT"]].tail(3))
```

```python
# 3. --- PREDICTIVE ALIGNMENT & FEATURE ENGINEERING ---
all_data = []
epsilon = 1e-9
price_years = sorted(df_prices.index.unique())

features = [
    'ROA', 'ROE', 'Operating_Margin', 'PE_Ratio', 'PS_Ratio', 'PB_Ratio',
    'Debt_to_Equity', 'Asset_Turnover', 'EPS_Growth', 'Company Market Cap',
    'Momentum_1Y'
]

print("Aligning Data (Target = Alpha)...")
for i in range(len(price_years) - 1):
    t_now = price_years[i]
    t_next = price_years[i + 1]

    # Calculate Raw Returns
    forward_returns = (df_prices.loc[t_next] / df_prices.loc[t_now]) - 1
    forward_returns = forward_returns.dropna()

    # Calculate Benchmark Return (Equal Weighted) for Alpha subtraction
    bench_ret = forward_returns.mean()

    f_snap = df_funda[df_funda['Year'] == (t_now - 1)].copy()

    # Feature Calculations
    f_snap['ROA'] = f_snap['Net Income'] / (f_snap['Total Assets, Reported'] + epsilon)
    f_snap['ROE'] = f_snap['Net Income'] / (f_snap['Total Equity'] + epsilon)
    f_snap['Operating_Margin'] = f_snap['Operating Income'] / (f_snap['Revenue'] + epsilon)
    f_snap['PE_Ratio'] = f_snap['Company Market Cap'] / (f_snap['Net Income'] + epsilon)
    f_snap['PS_Ratio'] = f_snap['Company Market Cap'] / (f_snap['Revenue'] + epsilon)
    f_snap['PB_Ratio'] = f_snap['Company Market Cap'] / (f_snap['Total Equity'] + epsilon)
    f_snap['Debt_to_Equity'] = f_snap['Total Debt'] / (f_snap['Total Equity'] + epsilon)
    f_snap['Asset_Turnover'] = f_snap['Revenue'] / (f_snap['Total Assets, Reported'] + epsilon)

    # YoY Growth
    df_prev = df_funda[df_funda['Year'] == (t_now - 2)]
    f_snap = f_snap.merge(df_prev[['Ticker', 'Net Income']], on='Ticker', how='left', suffixes=('', '_prev'))
    f_snap['EPS_Growth'] = (f_snap['Net Income'] - f_snap['Net Income_prev']) / (
                f_snap['Net Income_prev'].abs() + epsilon)

    # Momentum
    if (t_now - 1) in df_prices.index:
        mom = (df_prices.loc[t_now] / df_prices.loc[t_now - 1]) - 1
        f_snap['Momentum_1Y'] = f_snap['Ticker'].map(mom)
    else:
        f_snap['Momentum_1Y'] = 0

    # Outliers removal
    for feat in features:
        f_snap[feat] = f_snap[feat].clip(lower=f_snap[feat].quantile(0.01), upper=f_snap[feat].quantile(0.99))

    # Filling gaps in the data using the median to lessen the impact of extremes
    f_snap = f_snap.fillna(f_snap.median(numeric_only=True))
    if(i == 0):
        print(f_snap[["Ticker", "Year", "PB_Ratio", "PE_Ratio"]].head(10))

    for ticker, ret in forward_returns.items():
        row = f_snap[f_snap['Ticker'] == ticker]
        if not row.empty:
            combined = row.iloc[0].to_dict()
            # Target = market alpha
            combined['target_alpha'] = ret - bench_ret
            combined['actual_return'] = ret  # Keep for backtest visualization
            combined['buy_year'] = t_now
            all_data.append(combined)
```

```python
master_df = pd.DataFrame(all_data).replace([np.inf, -np.inf], np.nan).fillna(0)
print(master_df[["Ticker", "Year", "PB_Ratio", "PE_Ratio"]].head(5))
```

```python

# 4. --- XGBOOST WALK-FORWARD ---
results = []
available_years = sorted(master_df['buy_year'].unique())
train_window = 10

print("\nStarting Alpha Prediction Walk-Forward...")
print("-" * 65)
print(f"{'Year':<8} | {'D1 Alpha':<10} | {'MAE (%)':<10} | {'S&P (EW)':<10}")
print("-" * 65)

for i in range(train_window, len(available_years)):
    train_set = master_df[master_df['buy_year'].isin(available_years[i - train_window: i])]
    val_year = available_years[i - 1]
    val_set = master_df[master_df['buy_year'] == val_year]
    test_set = master_df[master_df['buy_year'] == available_years[i]]

    # MODIFICATION: Using target_alpha
    X_train, y_train = train_set[features], train_set['target_alpha']
    X_val, y_val = val_set[features], val_set['target_alpha']
    X_test = test_set[features]

    model = xgb.XGBRegressor(
        n_estimators=1000,
        learning_rate=0.01,
        max_depth=4,
        subsample=0.8,
        colsample_bytree=0.8,
        random_state=42,
        n_jobs=-1
    )

    model.fit(X_train, y_train, eval_set=[(X_val, y_val)], verbose=False)

    test_res = test_set[['Ticker', 'actual_return', 'target_alpha']].copy()
    test_res['pred_alpha'] = model.predict(X_test)

    if len(test_res) >= 30:
        # Calculate MAE in percent
        mae_pct = mean_absolute_error(test_res['target_alpha'], test_res['pred_alpha'])

        # Rank by predicted Alpha
        test_res['decile'] = pd.qcut(test_res['pred_alpha'].rank(method='first'), 10, labels=range(10, 0, -1))

        current_year = available_years[i]
        d1_actual_ret = test_res[test_res['decile'] == 1]['actual_return'].mean()
        bench_ret = test_res['actual_return'].mean()
        d1_alpha = d1_actual_ret - bench_ret

        print(f"{current_year:<8} | {d1_alpha:>9.2%} | {mae_pct:>9.2%} | {bench_ret:>9.2%}")

        y_res = {'Year': current_year, 'Bench': bench_ret, 'MAE': mae_pct}
        for d in range(1, 11):
            y_res[f'D{d}'] = test_res[test_res['decile'] == d]['actual_return'].mean()
        results.append(y_res)

# --- 5. FEATURE IMPORTANCE ---
print("\nPlotting Feature Importance for the final model...")

# Get the importance dictionary
importance = model.get_booster().get_score(importance_type='weight')

# Sort and plot
fig, ax = plt.subplots(figsize=(10, 8))
xgb.plot_importance(model, max_num_features=15, importance_type='weight', ax=ax, grid=False)
plt.title("Top 15 Fundamental Drivers of Alpha")
plt.show()
```

```python
# --- 5. FINAL SUMMARY & VISUALIZATION ---
perf = pd.DataFrame(results)
num_years = len(perf)

# Prepare data for the Bar Chart
decile_labels = [f'D{d}' for d in range(1, 11)]
cagrs = []

for d in range(1, 11):
    col = f'D{d}'
    perf[f'{col}_Cum'] = (1 + perf[col]).cumprod()
    mult = perf[f'{col}_Cum'].iloc[-1]
    cagr = (mult ** (1 / num_years)) - 1
    cagrs.append(cagr)

bench_mult = (1 + perf['Bench']).cumprod().iloc[-1]
bench_cagr = (bench_mult ** (1 / num_years)) - 1

# --- Plot 1: Bar Chart (Monotonicity) ---
plt.figure(figsize=(10, 6))
colors = ['#1f77b4' if c > bench_cagr else '#d62728' for c in cagrs] # Blue if beats bench, else red
plt.bar(decile_labels, cagrs, color=colors, alpha=0.8, label='Decile CAGR')
plt.axhline(bench_cagr, color='black', linestyle='--', label=f'Benchmark ({bench_cagr:.2%})')

plt.title("Annualized Return by Predicted Alpha Decile", fontsize=14)
plt.ylabel("CAGR (%)")
plt.xlabel("Decile (D1 = Highest Predicted Alpha)")

# Format Y-axis as percentages
import matplotlib.ticker as mtick
plt.gca().yaxis.set_major_formatter(mtick.PercentFormatter(1.0))
plt.legend()
plt.grid(axis='y', alpha=0.3)
plt.show()

# --- Plot 2: Cumulative Growth (Log Scale) ---
plt.figure(figsize=(10, 6))
plt.plot(perf['Year'], perf['D1_Cum'], label='Top Decile (D1)', lw=3, color='blue')
plt.plot(perf['Year'], (1 + perf['Bench']).cumprod(), label='Benchmark (EW)', color='black', ls='--')
plt.yscale('log')
plt.title("Growth of $1: D1 vs Benchmark")
plt.ylabel("Cumulative Multiple (Log Scale)")
plt.legend()
plt.grid(True, which="both", ls="-", alpha=0.2)
plt.show()
```
